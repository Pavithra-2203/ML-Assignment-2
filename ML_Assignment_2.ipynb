```python

# Step 1: Import the required libraries
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Step 2: Load the breast cancer dataset
cancer = load_breast_cancer()

# Step 3: Convert the dataset into a DataFrame
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)

# Step 4: Add the target column
df["diagnosis"] = [cancer.target_names[i] for i in cancer.target]

# Step 5: Save the dataset as a CSV file
df.to_csv("test_data.csv", index=False)

# Step 6: Display the dataset information
print("Dataset loaded successfully!")
print("Total records:", df.shape[0])
print("Total features:", len(cancer.feature_names))

# Display first 5 records
display(df.head())
```

```python

# STEP 3: Separate input features and target

from sklearn.model_selection import train_test_split

# X contains the 30 input features
X = df.drop("diagnosis", axis=1)

# y contains the target (0 = benign, 1 = malignant)
y = df["diagnosis"].map({"benign": 0, "malignant": 1})

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the results
print("Data splitting completed successfully!")
print("Total records:", len(df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Number of input features:", X_train.shape[1])

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python

# Create the test dataset with the target column
test_df = X_test.copy()

test_df["diagnosis"] = y_test.map({
    0: "benign",
    1: "malignant"
})

# Save the test dataset
test_df.to_csv("test_data.csv", index=False)

print("Test dataset saved successfully!")
print("Test dataset shape:", test_df.shape)
print("CSV file: test_data.csv")

display(test_df.head())
```

```python

# STEP 4: Import required ML algorithms

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

import joblib
import os

# Create a folder to save trained models
os.makedirs("model", exist_ok=True)

# STEP 5: Define all five models

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(max_iter=2000))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(n_neighbors=5))
    ]),

    "Naive Bayes": GaussianNB(),

    "Support Vector Machine": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ])
}

print("All five ML models are ready!")
```

```python

# STEP 6: Train all five ML models

for name, model in models.items():

    # Train the model using training data
    model.fit(X_train, y_train)

    print(name, "trained successfully!")

print("\nAll five models trained successfully!")
```

```python

# STEP 7: Evaluate all five models

results = []

for name, model in models.items():

    # Predict using test data
    y_pred = model.predict(X_test)

    # Get prediction probabilities for AUC
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate all six metrics
    accuracy = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    mcc = matthews_corrcoef(y_test, y_pred)

    # Store results
    results.append({
        "ML Model Name": name,
        "Accuracy": accuracy,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "MCC": mcc
    })

    print(name, "evaluation completed!")

# Convert results into a table
results_df = pd.DataFrame(results)

# Display comparison table
print("\nFINAL MODEL COMPARISON")
display(results_df.round(4))

# Save comparison results
results_df.to_csv("model_comparison.csv", index=False)

print("Model comparison saved successfully!")
```

```python

# STEP 8: Save all five trained models

for name, model in models.items():

    filename = name.lower().replace(" ", "_") + ".joblib"

    joblib.dump(model, os.path.join("model", filename))

    print(filename, "saved successfully!")

print("\nAll five models saved in the model folder!")
```
